```python
import os

# Logical CPU cores (includes hyperthreading)

print("Logical CPU cores:", os.cpu_count())

import psutil

import GPUtil

try:

    physical_cores = psutil.cpu_count(logical=False)

    logical_cores = psutil.cpu_count(logical=True)

    print(f"Physical cores: {physical_cores}")

    print(f"Logical cores: {logical_cores}")

except ImportError:

    print("psutil not installed. Run 'pip install psutil'")

## turn on when running on Lamuda work station, for CPU (-1), for GPU (0 or/and 1)

os.environ["CUDA_VISIBLE_DEVICES"] = "-1"  # use CPU only
 
import tensorflow as tf

devices = tf.config.list_physical_devices()

print(devices)

print("CPU", tf.config.list_physical_devices('CPU'))

print("GPU", tf.config.list_physical_devices('GPU'))
 
# === Static memory and GPU info ===

mem = psutil.virtual_memory()

print(f"total: {mem.total / 1e9:.2f} GB")

print(f"used: {mem.used / 1e9:.2f} GB")

print(f"available: {mem.available / 1e9:.2f} GB")
 
gpus = GPUtil.getGPUs()

for gpu in gpus:

    print(f"GPU ID {gpu.id}: {gpu.name}")

    print(f"  Total Memory: {gpu.memoryTotal} MB")

    print(f"  Used Memory:  {gpu.memoryUsed} MB")

    print(f"  Free Memory:  {gpu.memoryFree} MB")
 
# === Monitor CPU and memory usage in background ===

import threading

import shutil

import time

import datetime

def monitor_resources(stop_event, interval=1.0):

    while not stop_event.is_set():

        mem = psutil.virtual_memory()

        cpu = psutil.cpu_percent()

        used = mem.used / (1024 ** 3)

        total = mem.total / (1024 ** 3)

        current_time = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")

        line = f"[{current_time}][RESOURCE] CPU: {cpu:5.1f}% | Memory: {used:.1f} / {total:.1f} GB used"

        cols = shutil.get_terminal_size((80, 20)).columns

        print(line.ljust(cols), end="\r")

        time.sleep(interval)
 
```
